```python
# Import necessary libraries
import pandas as pd
from datetime import datetime
from IPython.display import display, Markdown

todays_date = datetime.now().strftime("%Y-%m-%d")
student_id = "Lasgat0980"

# Load the Breast Cancer dataset
data = pd.read_csv(r'C:\Users\ldick\Downloads\breastCancerDataset.csv')

# Display the first few rows to understand what the data looks like
display(Markdown("### Step 1: Load the Dataset"))
display(Markdown(f"Date: {todays_date}"))
display(Markdown(f"Student ID: {student_id}"))
display(data.head())
```

```python
display(Markdown("### Step 2: Initial Exploration of the Dataset"))
display(Markdown(f"Date: {todays_date}"))
display(Markdown(f"Student ID: {student_id}"))

# Display dataset information
data.info()

# Display summary statistics
display(data.describe())
```

```python
display(Markdown("### Step 3: Check for Missing Values"))
display(Markdown(f"Date: {todays_date}"))
display(Markdown(f"Student ID: {student_id}"))

# Check for missing values
print(data.isnull().sum())
```

```python
display(Markdown("### Step 4: Check for Duplicate Rows"))
display(Markdown(f"Date: {todays_date}"))
display(Markdown(f"Student ID: {student_id}"))

# Check for duplicate rows
print("Number of duplicate rows:", data.duplicated().sum())

# Remove duplicate rows
data.drop_duplicates(inplace=True)

# Verify duplicates have been removed
print("Number of duplicate rows after cleaning:", data.duplicated().sum())
```

```python
display(Markdown("### Step 5: Standardize Features"))
display(Markdown(f"Date: {todays_date}"))
display(Markdown(f"Student ID: {student_id}"))

# Import StandardScaler
from sklearn.preprocessing import StandardScaler

# Standardize concavity1 and concavity3
scaler = StandardScaler()
data[['concavity1', 'concavity3']] = scaler.fit_transform(data[['concavity1', 'concavity3']])

# Display the standardized features
display(data[['concavity1', 'concavity3']].head())
```

```python
display(Markdown("### Step 6: Split the Dataset"))
display(Markdown(f"Date: {todays_date}"))
display(Markdown(f"Student ID: {student_id}"))

# Import train_test_split
from sklearn.model_selection import train_test_split

# Separate features and target variable
X = data.drop(columns=['ID', 'Diagnosis'])
y = data['Diagnosis']

# Split the dataset into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)

# Display the shapes of the training and testing sets
print("Training set shape:", X_train.shape)
print("Testing set shape:", X_test.shape)
```

```python
display(Markdown("### Step 7: Train and Evaluate the SVM Model"))
display(Markdown(f"Date: {todays_date}"))
display(Markdown(f"Student ID: {student_id}"))

# Import necessary libraries
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, classification_report

# Train the SVM model
svm_model = SVC()
svm_model.fit(X_train, y_train)

# Make predictions
y_pred_svm = svm_model.predict(X_test)

# Evaluate the model
print("SVM Accuracy:", accuracy_score(y_test, y_pred_svm))
print("SVM Classification Report:")
print(classification_report(y_test, y_pred_svm))
```

```python
display(Markdown("### Step 8: Hyperparameter Tuning with GridSearchCV"))
display(Markdown(f"Date: {todays_date}"))
display(Markdown(f"Student ID: {student_id}"))

# Import GridSearchCV
from sklearn.model_selection import GridSearchCV

# Define the parameter grid
param_grid = {
    'C': [0.1, 1, 10, 100],
    'kernel': ['linear', 'rbf', 'poly']
}

# Perform hyperparameter tuning
grid_search = GridSearchCV(SVC(), param_grid, cv=5)
grid_search.fit(X_train, y_train)

# Display the best parameters
print("Best Parameters:", grid_search.best_params_)

# Make predictions using the best model
best_svm_model = grid_search.best_estimator_
y_pred_svm_tuned = best_svm_model.predict(X_test)

# Evaluate the tuned model
print("Tuned SVM Accuracy:", accuracy_score(y_test, y_pred_svm_tuned))
print("Tuned SVM Classification Report:")
print(classification_report(y_test, y_pred_svm_tuned))
```

```python
display(Markdown("### Step 9: Train and Evaluate the XGBoost Model"))
display(Markdown(f"Date: {todays_date}"))
display(Markdown(f"Student ID: {student_id}"))

# Import necessary libraries
from xgboost import XGBClassifier

# Convert diagnosis labels to numerical values
y_train_xgb = y_train.map({'B': 0, 'M': 1})
y_test_xgb = y_test.map({'B': 0, 'M': 1})

# Train the XGBoost model
xgb_model = XGBClassifier(random_state=42)
xgb_model.fit(X_train, y_train_xgb)

# Make predictions
y_pred_xgb = xgb_model.predict(X_test)

# Evaluate the model
print("XGBoost Accuracy:", accuracy_score(y_test_xgb, y_pred_xgb))
print("XGBoost Classification Report:")
print(classification_report(y_test_xgb, y_pred_xgb))
```

```python
display(Markdown("### Step 10: Hyperparameter Tuning for XGBoost"))
display(Markdown(f"Date: {todays_date}"))
display(Markdown(f"Student ID: {student_id}"))

# Define the parameter grid
param_grid_xgb = {
    'n_estimators': [50, 100],
    'learning_rate': [0.1, 0.2],
    'max_depth': [3, 4]
}

# Perform hyperparameter tuning
grid_search_xgb = GridSearchCV(XGBClassifier(random_state=42), param_grid_xgb, cv=5)
grid_search_xgb.fit(X_train, y_train_xgb)

# Display the best parameters
print("Best Parameters:", grid_search_xgb.best_params_)

# Make predictions using the best model
best_xgb_model = grid_search_xgb.best_estimator_
y_pred_xgb_tuned = best_xgb_model.predict(X_test)

# Evaluate the tuned model
print("Tuned XGBoost Accuracy:", accuracy_score(y_test_xgb, y_pred_xgb_tuned))
print("Tuned XGBoost Classification Report:")
print(classification_report(y_test_xgb, y_pred_xgb_tuned))
```
